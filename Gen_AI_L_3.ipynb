```python
from pathlib import Path
import random
import torch
import matplotlib.pyplot as plt

SEED = 42
random.seed(SEED)
torch.manual_seed(SEED)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

out = Path("lab_outputs")
out.mkdir(exist_ok=True)

print("Device:", device)
```

```python
from transformers import AutoTokenizer, AutoModelForCausalLM

model_id = "openai-community/gpt2"

tokenizer = AutoTokenizer.from_pretrained(model_id)

tokenizer.pad_token = tokenizer.eos_token

model = AutoModelForCausalLM.from_pretrained(
    model_id
).to(device)

model.eval()
```

```python
def continue_text(
    prompt,
    sample=False,
    temperature=0.8,
    seed=42
):

    torch.manual_seed(seed)

    inputs = tokenizer(
        prompt,
        return_tensors="pt"
    ).to(device)

    options = {
        "do_sample": sample,
        "max_new_tokens": 45,
        "pad_token_id": tokenizer.eos_token_id
    }

    if sample:
        options.update(
            temperature=temperature,
            top_p=0.9
        )

    with torch.no_grad():

        tokens = model.generate(
            **inputs,
            **options
        )

    return tokenizer.decode(
        tokens[0, inputs.input_ids.size(1):],
        skip_special_tokens=True
    ).strip()


prompts = [
    "A story about a robot begins like this:",
    "In a school library, a tiny robot found a lost book. It"
]

for prompt in prompts:

    print("\nPROMPT:", prompt)

    print("GREEDY:", continue_text(prompt))

    for temp in (0.6, 1.0):

        for seed in (42, 43):

            print(
                temp,
                seed,
                continue_text(
                    prompt,
                    sample=True,
                    temperature=temp,
                    seed=seed
                )
            )
```

```python
examples = [
    ("graphs", "Sketch a graph, then mark a path through three vertices."),
    ("Python loops", "Trace a loop by writing the counter after each step."),
    ("fractions", "Draw equal parts and label the fraction you shaded."),
    ("sets", "Draw a Venn diagram and label the intersection."),
    ("sorting", "Sort five values by hand and record each swap."),
    ("probability", "List the outcomes of two coin tosses before counting.")
]


tuned = AutoModelForCausalLM.from_pretrained(
    model_id
).to(device)

tuned.requires_grad_(False)

tuned.transformer.h[-1].requires_grad_(True)

opt = torch.optim.AdamW(
    tuned.transformer.h[-1].parameters(),
    lr=2e-5
)


for epoch in range(3):

    tuned.train()

    for topic, answer in examples:

        prefix = tokenizer.encode(
            f"Topic: {topic}\nStudy tip:"
        )

        target = tokenizer.encode(
            " " + answer
        ) + [tokenizer.eos_token_id]

        ids = torch.tensor(
            [prefix + target],
            device=device
        )

        labels = torch.tensor(
            [[-100] * len(prefix) + target],
            device=device
        )

        opt.zero_grad()

        loss = tuned(
            input_ids=ids,
            labels=labels
        ).loss

        loss.backward()

        opt.step()

    print(
        "Epoch:",
        epoch + 1,
        "last example loss:",
        round(loss.item(), 3)
    )


tuned.save_pretrained(out / "study_tip_gpt2")

tokenizer.save_pretrained(out / "study_tip_gpt2")
```

```python
held_out = [
    "recursion",
    "matrices",
    "binary search"
]

base_model = model

for topic in held_out:

    prompt = f"Topic: {topic}\nStudy tip:"

    model = base_model

    print(
        topic,
        "base:",
        continue_text(prompt)
    )

    model = tuned.eval()

    print(
        topic,
        "tuned:",
        continue_text(prompt)
    )

model = base_model
```
